```python
#Variable numérica de Potencial de compra
df['Potencial_num'] = df['Potencial de compra'].replace('Sin registro', np.nan)

df['Potencial_num'] = df['Potencial_num'].str.replace('%', '', regex=False)

df['Potencial_num'] = pd.to_numeric(df['Potencial_num'], errors='coerce')
```

```python
df['Potencial_num'].describe()
```

```python
n = df['Potencial_num'].count()
n
```

```python
#Límites
Max = df['Potencial_num'].max()
Min = df['Potencial_num'].min()

Limites = [Min, Max]
Limites
```

```python
R = Max-Min
R
```

```python
#regla de Sturges
ni = 1+3.32*np.log10(n)
ni
```

```python
i = R/10
i
```

```python
intervalos=np.linspace(Min, Max, 11)
intervalos
```

```python
#Creamos las categorías
categorias=[
    "20%-28%",
    "28%-36%",
    "36%-44%",
    "44%-52%",
    "52%-60%",
    "60%-68%",
    "68%-76%",
    "76%-84%",
    "84%-92%",
    "92%-100%"
]
```

```python
#Creamos nuevamente la variable numérica
df['Potencial_num'] = df['Potencial de compra'].str.replace('%', '', regex=False)

df['Potencial_num'] = pd.to_numeric(
    df['Potencial_num'],
    errors='coerce'
)
```

```python
#Finalmente creamos las categorías en la columna numérica
df['Potencial_num']=pd.cut(
    x=df['Potencial_num'],
    bins=intervalos,
    labels=categorias,
    include_lowest=True
)

df['Potencial_num']
```

```python
Tabla_freq=df['Potencial_num'].value_counts(sort=False).reset_index()
Tabla_freq
```

```python
Filtro=Tabla_freq[Tabla_freq['count']>0]
Filtro
```

```python
#Ajustamos el índice
Filtro_index=Filtro.set_index('Potencial_num')
Filtro_index
```

```python
Filtro_index.plot(
    kind='bar',
    width=0.8,
    figsize=(12,4),
    color="purple",
    rot=45
)

plt.title('Potencial de Compra por Categorías')
plt.xlabel('Rango de Potencial de Compra')
plt.ylabel('Frecuencia')
```
